**a) Defina con sus palabras Sistema, Modelo, Simulación y Solución Analítica:**

- **Sistema:**
Un sistema es un conjunto de elementos ya sean recursos, procesos, entre otros, que interactúan entre sí y con el entorno para cumplir un objetivo. Por ejemplo puede ser físico como una fábrica, abstracto como en TI un algoritmo, natural como un ecosistema.

- **Modelo:**
Un modelo es una representación simple y estructurada de un sistema real el cual se encarga de captura los aspectos más importantes para analizarlo, predecir su comportamiento o incluso tomar decisiones.

- **Simulación:**
Una simulación es una técnica que permite experimentar con un modelo, generalmente se realizar por medio de un computador, para observar su comportamiento en distintos escenarios y condiciones que se planteen. Para ser exacto no busca resolver el sistema como tal de forma exacta, sino que aproxima y analiza cómo evolucionaría con diferentes parámetros o eventos dados.

- **Solución Analítica:**
Una Solución Analítica es el resultado obtenido aplicando métodos matemáticos exactos ya se mediante álgebra, cálculo, probabilidad, ecuaciones diferenciales, entre otros metodos matematicos. Estos mecanismos son aplicados al modelo del sistema.

**b) Dé ejemplos de sistemas para cada par y justifique:**

- **Estático vs. Dinámico**

**Estático:** Un sistema para control de calidad el cual mide el diámetro de un lote de tornillos(X). El modelo solo analiza una foto generada en el instante, **sin evolución temporal**.

**Dinámico:** El sistema de bicicletas compartidas (Robledo - C4TA), donde las variables cambian con el tiempo: disponibilidad de bicicletas por ejemplo.

**Justificación:** Lo estático no depende del tiempo; lo dinámico sí.

- **Determinista vs. Estocástico**

**Determinista:** El cálculo de la trayectoria de un proyectil en condiciones ideales, sin viento ni resistencia del aire. Conociendo la velocidad inicial y el ángulo de lanzamiento, las ecuaciones de movimiento siempre dan el mismo resultado.

**Estocástico:** El lanzamiento de una moneda o la llegada de clientes a una estación de bicicletas. Aunque las probabilidades son conocidas, el resultado es aleatorio.

**Justificación:** En un modelo determinista no hay incertidumbre; en uno estocástico interviene el azar.

- **Discreto vs. Continuo vs. Híbrido**

**Discreto:** La cola de un banco, donde las variables cambian en pasos definidos, número de clientes que llega o se va.

**Continuo:** El crecimiento de la población mundial modelado por ecuaciones diferenciales, cambia en forma continua con el tiempo.

**Híbrido:** El tráfico vehicular en una ciudad. Hay un flujo continuo de autos, pero también eventos discretos como la llegada de un semáforo en rojo.

**Justificación:** Discreto = cambios por eventos puntuales; Continuo = evolución en el tiempo; Híbrido = combinación de ambos.

- **Abierto vs. Cerrado**

**Abierto:** Un hospital, que recibe pacientes, brinda servicios y los egresa. Intercambia recursos, información y personas con su entorno.

**Cerrado:** Una máquina de café automática con una cantidad fija de tazas, sin interacción continua con el exterior durante su operación.

**Justificación:** Abierto interactúa con el entorno; Cerrado no.

- **Estocástico:** Las llegadas de tickets no siguen un patrón exacto y determinista, sino que presentan aleatoriedad. Lo mismo ocurre con los tiempos de servicio, que varían de un ticket a otro.

- **Discreto:** Los cambios en el sistema no ocurren de manera continua, si no en momentos concretos donde suceden eventos puntuales: la llegada de un nuevo ticket, el inicio de atención por parte de un agente, o la finalización del servicio.

- **Cerrado:** El sistema arranca con un conjunto finito y fijo de 1 000 peticiones y no recibe más llegadas desde el entorno.

**4. Inventario en una bodega con reabastecimiento discreto mensual y consumo continuo**

- **Estocástico:** El consumo varía debido a la incertidumbre en la demanda de clientes o áreas internas, lo que hace que los niveles de inventario no sean completamente predecibles.

- **Abierto:** El sistema no está aislado. Interactúa con su entorno en dos sentidos: por un lado, el inventario se reduce debido al consumo y por otro lado, se incrementa gracias al reabastecimiento que proviene de proveedores.

## **P4 — Crecimiento poblacional — 30%**

**Objetivo:** Partiendo del ejemplo de crecimiento poblacional presentado en la **Clase 6** (crecimiento constante o “lineal en el tiempo”), extienda dicho ejemplo a un **crecimiento cuadrático** proporcional a la población. Construya el código para este caso y compárelo con las predicciones obtenidas en el modelo de crecimiento constante (Clase 6).

**Datos:** Use los **mismos datos** de estimaciones de población mundial empleados en el notebook CLase6 (serie **US Census** o **UN**). Documente brevemente cualquier decisión (columnas, años usados).

#### **Definición matematica de modelo lineal/constante (Construido en Clase 6)**

* Crecimiento neto constante $G_t$ (adición anual fija). Para cada año $t$:

  $$
  \widehat{P}_{t+1} \;=\; P_t \;+\; G_t,
  \quad \text{con}\quad G_t=\text{promedio de }(P_{t+1}-P_t).
  $$


#### **Definición matematica de modelo de crecimiento cuadrático**

* El **crecimiento neto** depende **cuadráticamente** de la población:

  $$
  G_t \;=\; \alpha\,P_t \;+\; \beta\,P_t^{2}.
  $$
* **Estimación.** $\alpha,\beta$ son parametreos ajustables, tome $\alpha = 0.025$ y $\beta = -0.0018$.

* **Tareas.**

  1. **Construya el código** que le permita hacer las predicciones con un crecimiento cuadrático.
  2. **Calcule** predicciones.
  3. Evalúe el **error absoluto medio** y el **error relativo medio** (con los mismos años que en el modelo lineal) y realice una tabla comparativa (**lineal vs. cuadrático**)
  4. Gráfique: **Tiempo vs. Población:** para la serie **real**, las estimación **lineal** y la estimación **cuadrática**.
  5. En 3 o 4 líneas **concluya** cuál modelo se ajusta mejor y por qué.

```python
# ===============================
# Crecimiento poblacional
# ===============================
# Objetivo:
# Comparar modelo lineal (Clase 6) vs modelo cuadrático (nuevo).
# Fuente de datos: US Census.
# Años usados: 1950–2016. Valores expresados en miles de millones.
# ===============================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pandas import read_html
from os.path import basename, exists

# Descargar datos
def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve
        local, _ = urlretrieve(url, filename)
        print('Downloaded ' + local)

download('https://raw.githubusercontent.com/AllenDowney/'
         'ModSimPy/master/data/World_population_estimates.html')

# Cargar tablas
filename = 'World_population_estimates.html'
tables = read_html(filename, header=0, index_col=0, decimal='M')

table2 = tables[2]
table2.columns = ['census', 'prb', 'un', 'maddison',
                  'hyde', 'tanton', 'biraben', 'mj',
                  'thomlinson', 'durand', 'clark']

# Serie US Census en miles de millones
census = table2.census / 1e9

# ===============================
# MODELO LINEAL (Clase 6)
# ===============================
t_0 = census.index[0]       # año inicial (1950)
t_end = census.index[-1]    # año final (2016)
elapsed_time = t_end - t_0

p_0 = census[t_0]
p_end = census[t_end]

# Crecimiento promedio anual
annual_growth = (p_end - p_0) / elapsed_time

linear_results = pd.Series(index=range(t_0, t_end+1), dtype=float)
linear_results[t_0] = p_0
for t in range(t_0, t_end):
    linear_results[t+1] = linear_results[t] + annual_growth

# ===============================
# MODELO CUADRÁTICO
# ===============================
# Gt = alpha*Pt + beta*Pt^2
alpha, beta = 0.025, -0.0018

quad_results = pd.Series(index=range(t_0, t_end+1), dtype=float)
quad_results[t_0] = p_0
for t in range(t_0, t_end):
    Pt = quad_results[t]
    Gt = alpha*Pt + beta*(Pt**2)
    quad_results[t+1] = Pt + Gt

# ===============================
# EVALUACIÓN DE ERRORES
# ===============================
# (Mean Absolute Error)
def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

# (Mean Absolute Percentage Error)
def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred)/y_true)) * 100

census_aligned = census.loc[t_0:t_end]

tabla = pd.DataFrame({
    "Modelo": ["Lineal", "Cuadrático"],
    "Error Absoluto": [
        mae(census_aligned, linear_results),
        mae(census_aligned, quad_results)
    ],
    "Error relativo": [
        mape(census_aligned, linear_results),
        mape(census_aligned, quad_results)
    ]
})

print("Tabla Comparación de Errores")
print(tabla)

# ===============================
# GRÁFICA
# ===============================
plt.figure(figsize=(9,5))
census.plot(style="o-", label="US Census")
linear_results.plot(style="--", label="Modelo lineal")
quad_results.plot(style=":", label="Modelo cuadrático")

plt.title("Crecimiento poblacional mundial (1950–2016)")
plt.xlabel("Año")
plt.ylabel("Población (miles de millones)")
plt.legend()
plt.grid(True)
plt.show()
```

### Conclusiones
El análisis de errores muestra que el modelo cuadrático presenta un ajuste superior al modelo lineal, con valores menores tanto de MAE como de MAPE.
Esto significa que, en promedio, el cuadrático se equivoca menos tanto en magnitud como en proporción relativa.
Además, la forma del cuadrático capta mejor la tendencia de desaceleración en el crecimiento poblacional, mientras que el modelo lineal tiende a sobrestimar los valores en los años recientes.
